### TUPLES

```python
a =(10,20,30) #Concatenate the two tuples and store it in “t_combine”
b =(40,50,60)
t_combine = a+b
print(t_combine)
```

```python
print(t_combine*3) #Repeat the elements of “t_combine” 3 times
```

```python
print(t_combine[2]) #Access the 3rd element from “t_combine”
```

```python
print(t_combine[:3]) #Access the first three elements from “t_combine”
```

```python
print(t_combine[-3:]) #Access the last three elements from “t_combine”
```
